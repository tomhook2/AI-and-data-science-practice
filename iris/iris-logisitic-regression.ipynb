# Iris Dataset - Logistic Regression

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
```

```python
df_iris = sns.load_dataset('iris')
```

```python
X = df_iris.drop('species', axis=1)
y = df_iris.species

model = LogisticRegression(max_iter=1000)
model.fit(X, y)

model
```

```python
y_true = df_iris.species
y_pred = model.predict(X)

print(classification_report(y_true, y_pred))
```

Predictions and confidence for each row

```python
# Extract probabilities from the model and take the maximum one as that will equate to the confidence in the prediction (predicts the label for the highest value)
proba = model.predict_proba(X)
confidence = proba.max(axis=1)

# Build results table
results = pd.DataFrame({
    "y_true": y_true.values,
    "y_pred": y_pred,
    "confidence": confidence
})

print(results.head(3))
print(results.tail(3))
```
